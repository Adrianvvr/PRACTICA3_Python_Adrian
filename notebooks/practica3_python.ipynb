## Ejercicio 1 — Catálogo de productos como lista

```python
# Partimos del catálogo base
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# 1. Total de elementos y deducción de productos
# Como la estructura intercala 1 nombre y 1 precio por producto, la cantidad real de artículos es la mitad de la longitud total.
total_elementos = len(catalogo)
num_productos = total_elementos // 2
print("1. Total elementos:", total_elementos, "-> Productos reales:", num_productos)

# 2. Separar nombres y precios mediante slicing
# Extraemos saltando de 2 en 2. Los nombres empiezan en la primera posición (0) y los precios en la segunda (1).
nombres = catalogo[0::2]
precios = catalogo[1::2]
print("\n2. Lista de nombres:", nombres)
print("2. Lista de precios:", precios)
print("   Comprobación precios (debe sumar 914.40):", sum(precios))

# 3. Obtener el precio del "Robot Aspirador" dinámicamente
# Evitamos escribir el índice a mano ('hardcoding') por si la lista cambia de orden en el futuro.
# Sabemos por la estructura de negocio que el precio siempre es la posición inmediatamente posterior al nombre (+1).
idx_robot = catalogo.index("Robot Aspirador")
precio_robot = catalogo[idx_robot + 1]
print("\n3. Precio del Robot Aspirador:", precio_robot, "€")

# 4. Últimos tres productos con índices negativos
# Consultamos la cola de la lista usando -3 para traer desde el antepenúltimo elemento hasta el final.
ultimos_tres_nombres = nombres[-3:]
print("\n4. Últimos tres nombres:", ultimos_tres_nombres)

# 5. Precio medio del catálogo
# Calculamos el promedio sumando el total y dividiendo por las líneas de producto, redondeando para formato moneda.
precio_medio = round(sum(precios) / len(precios), 2)
print("\n5. Precio medio del catálogo:", precio_medio, "€")
```

## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

# 1. Producto de la cuarta fila
# Navegamos a la fila 4 (índice 3) y luego a la columna producto (índice 1).
producto_cuarta = inventario[3][1]
print("1. Producto de la cuarta fila:", producto_cuarta)

# 2. Lista valor_stock con bucle for
# Transformamos el inventario creando una nueva estructura enfocada en el valor económico.
valor_stock = []
for fila in inventario:
    producto = fila[1]
    valor_linea = round(fila[2] * fila[3], 2)
    valor_stock.append([producto, valor_linea])

print("\n2. Valor stock por línea:")
for item in valor_stock:
    print("   ", item)

# 3. Valor inmovilizado total
# Creamos una variable acumuladora en 0 y le sumamos el valor de cada producto
valor_total = 0
for item in valor_stock:
    valor_total = valor_total + item[1]

print("\n3. Valor inmovilizado total:", valor_total, "€")


# 4. Slicing contiguo para ALM-SUR
# Extraemos la sublista asumiendo agrupación visual
inventario_sur = inventario[2:4]

# Calculamos el valor de este almacén paso a paso
valor_sur = 0
for fila in inventario_sur:
    unidades = fila[2]
    coste = fila[3]
    valor_sur = valor_sur + (unidades * coste)

# Redondeamos al final
valor_sur = round(valor_sur, 2)
print("4. Valor inmovilizado ALM-SUR:", valor_sur, "€")
```

Si los registros de un mismo almacén dejan de estar en posiciones consecutivas, el slicing extraería filas incorrectas basándose únicamente en su orden numérico y no en su contenido. Por ello, el slicing no es una estrategia de filtrado robusta; en entornos reales se deben utilizar condiciones lógicas que evalúen el dato de cada fila sin depender de su ubicación.

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00]

# 1. Asignación directa y modificación (el error típico)
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90
print("1. Lista original :", precios_originales)
print("   Lista rebajada :", precios_rebajados)
print("   Explicación: Al usar '=', no creamos una lista nueva, solo una 'etiqueta' que apunta a la misma lista. Modificar una altera la otra.")

# 2. Restaurar original y crear copias reales e independientes
precios_originales[0] = 59.90  # Restauramos el valor inicial

copia_con_list = list(precios_originales)
copia_con_slicing = precios_originales[:]

# Demostramos que son independientes
copia_con_list[0] = 10.00
print("\n2. Original tras cambiar la copia:", precios_originales)
print("   Copia independiente cambiada:", copia_con_list)

# 3. Rebaja del 10% iterando por posiciones (índices)
for i in range(len(copia_con_slicing)):
    nuevo_precio = copia_con_slicing[i] * 0.90
    copia_con_slicing[i] = round(nuevo_precio, 2)
    
print("\n3. Copia con 10% de rebaja:", copia_con_slicing)

# 4. Añadir e insertar productos en la lista rebajada
copia_con_slicing.extend([99.90, 120.00])
copia_con_slicing.insert(2, 75.50)
print("\n4. Tras añadir e insertar nuevos precios:", copia_con_slicing)

# 5. Eliminar elementos
del copia_con_slicing[4]
descartado = copia_con_slicing.pop()
print("\n5. Lista tras borrar posición 4 y el último:", copia_con_slicing)
print("   Valor extraído y guardado:", descartado)

# 6. Comprobar que viven en espacios de memoria distintos
print("\n6. Identificadores en memoria (id):")
print("   ID Original      :", id(precios_originales))
print("   ID Copia list()  :", id(copia_con_list))
print("   ID Copia slicing :", id(copia_con_slicing))
```

## Ejercicio 5 — Funciones propias para el cálculo de importes

```python
# 1 y 4. Definir calcular_importe con docstring
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el importe neto de un pedido aplicando un descuento porcentual.
    Devuelve el importe final redondeado a dos decimales.
    """
    bruto = unidades * precio_unitario
    neto = bruto * (1 - (descuento_pct / 100))
    return round(neto, 2)

# 2 y 4. Definir resumen_pedido con docstring
def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el desglose de un pedido.
    Devuelve una tupla con (importe_bruto, ahorro, importe_neto).
    """
    bruto = unidades * precio_unitario
    ahorro = bruto * (descuento_pct / 100)
    neto = bruto - ahorro
    return (round(bruto, 2), round(ahorro, 2), round(neto, 2))

# Mostrar la ayuda de la función
help(calcular_importe)

# 3. Llamada por palabra clave y desempaquetado de tupla
bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print("Desglose -> Bruto:", bruto, "€ | Ahorro:", ahorro, "€ | Neto:", neto, "€")

# 5. Cálculo sobre la lista de pedidos
pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = []
for ped in pedidos:
    # ped[0] son unidades, ped[1] es precio, ped[2] es descuento
    importe_calculado = calcular_importe(ped[0], ped[1], ped[2])
    importes.append(importe_calculado)

print("\nLista de importes :", importes)
print("Número de pedidos :", len(importes))
print("Importe máximo    :", max(importes), "€")
print("Importe mínimo    :", min(importes), "€")
print("Importes ordenados:", sorted(importes))
```

```python
skus_crudos = [
    "  elc-0012-es ", "HOG-0045-ES", "dep-0003-pt  ",
    "ofi-0021-es", "  ELC-0012-ES", "hog-0099-fr ",
]

# 1. Limpiar espacios y convertir a mayúsculas
skus_limpios = []
for sku in skus_crudos:
    skus_limpios.append(sku.strip().upper())
print("1. SKUs limpios:", skus_limpios)

# 2. Contar apariciones de "ELC-0012-ES"
conteo_elc = skus_limpios.count("ELC-0012-ES")
print("\n2. Apariciones de ELC-0012-ES:", conteo_elc)

# 3. Obtener posición de "OFI-0021-ES"
pos_ofi = skus_limpios.index("OFI-0021-ES")
print("3. Posición de OFI-0021-ES:", pos_ofi)

# 4. Separar componentes (familia, número, país)
skus_separados = []
for sku in skus_limpios:
    skus_separados.append(sku.split("-"))
print("\n4. SKUs separados:")
for partes in skus_separados:
    print("   ", partes)

# 5. Filtrar por país "ES" y reemplazar guiones por puntos
skus_es = []
for sku in skus_limpios:
    if sku.endswith("-ES"):
        skus_es.append(sku.replace("-", "."))
print("\n5. SKUs de España modificados:", skus_es)

# 6. Unir en una cadena única separada por " | "
cadena_unica = " | ".join(skus_es)
print("\n6. Cadena única unida:", cadena_unica)
```

## Ejercicio 8 — Del cálculo con listas al cálculo vectorizado

```python
import numpy as np

unidades = [3, 1, 5, 2, 4, 6, 2, 8, 3, 1]
precios  = [59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00]

# 1. Intento de multiplicar listas de Python
try:
    fallo = unidades * precios
except Exception as e:
    print("1. Error al multiplicar listas:", type(e).__name__, "-", e)

# 2. Convertir a arrays de NumPy y ver atributos
np_unidades = np.array(unidades)
np_precios = np.array(precios)

print("\n2. np_unidades -> shape:", np_unidades.shape, "| dtype:", np_unidades.dtype)
print("   np_precios  -> shape:", np_precios.shape, "| dtype:", np_precios.dtype)

# 3. Cálculo vectorizado (aquí NumPy sí sabe multiplicar elemento a elemento)
importes = np_unidades * np_precios
print("\n3. Array de importes:", importes)
print("   Suma total:", importes.sum(), "€")

# 4. Recargo logístico del 4% en una sola operación
importes_recargo = importes * 1.04
print("\n4. Importes con recargo:", importes_recargo)

# 5. Array mixto
mixto = np.array([1, "dos", 3.0, True])
print("\n5. Array mixto:", mixto)
print("   dtype del array mixto:", mixto.dtype)

# 6. Comparación de rendimiento usando comandos mágicos de Jupyter

print("Tiempo con listas estándar de Python:")
%timeit [u * p for u, p in zip(unidades, precios)]

print("\nTiempo vectorizado con NumPy:")
%timeit np_unidades * np_precios
```

**1. Error con listas:** El `TypeError` ocurre porque el operador `*` en listas nativas de Python no sabe multiplicar elementos entre sí; solo sirve para repetir una lista entera por un número entero.

**5. Array mixto y homogeneidad:** El `dtype` resultante es una cadena de texto (tipo `<U32` o similar) porque NumPy exige que todos los elementos de un array sean del mismo tipo para ser ultra-rápido en memoria, y el texto es el único tipo que puede englobar números, letras y booleanos sin perder información.

## Ejercicio 9 — Subsetting con máscaras booleanas

```python
# 1. Crear e imprimir la máscara booleana
caros = np_precios > 100
print("1. Máscara 'caros':")
print("  ", caros)

# 2. Extraer los precios que superan 100 €
precios_caros = np_precios[caros]
print("\n2. Precios superiores a 100 €:", precios_caros)

# 3. Contar cuántos productos cumplen la condición
cantidad_caros = caros.sum()
print("\n3. Cantidad de productos caros:", cantidad_caros)

# 4. Aplicar la misma máscara al array de importes y sumar
importes_caros = importes[caros]
suma_importes_caros = importes_caros.sum()
print("\n4. Importes de las líneas caras:", importes_caros)
print("   Suma de estos importes:", suma_importes_caros, "€")

# 5. Obtener las posiciones (índices) exactas donde se cumple la condición
posiciones_caros = np.where(caros)
print("\n5. Posiciones de los productos caros:", posiciones_caros[0])

# 6. Combinar múltiples condiciones con & (ambas deben cumplirse)
varias = np_unidades >= 2
seleccion = (np_precios > 100) & (np_unidades >= 2)

print("\n6. Importes de productos >100€ Y unidades >=2:")
print("  ", importes[seleccion])
```

Sumar booleanos devuelve un número porque, internamente en Python y NumPy, el valor `True` equivale matemáticamente a `1` y el valor `False` a `0`; por tanto, sumar una máscara booleana es la forma más eficiente de contar cuántas veces se cumple una condición.

## Ejercicio 10 — Matriz de ventas por trimestre y región

```python
import numpy as np

ventas_2d = np.array([
    [15200.50, 11800.25, 12950.00,  9870.40, 13400.10], # T1
    [16840.75, 12310.60, 13480.90, 11020.35, 14100.80], # T2
    [12535.72, 10663.06, 11221.59, 12673.08, 12334.38], # T3
])

# 1. Propiedades básicas de la matriz
print("1. Propiedades:")
print("   Shape:", ventas_2d.shape)
print("   Ndim :", ventas_2d.ndim)
print("   Size :", ventas_2d.size)

# 2. Extracciones completas (fila y columna)
fila_t2 = ventas_2d[1, :]
col_centro = ventas_2d[:, 4]
print("\n2. Fila T2:", fila_t2)
print("   Columna Centro:", col_centro)

# 3. Intersección 2D exacta (T3, Este)
# T3 es la fila 2, Este es la columna 2
valor_t3_este = ventas_2d[2, 2]
print("\n3. Valor T3 en el Este:", valor_t3_este, "€")

# 4. Slicing de submatriz (T1 y T2 para Norte, Sur, Este)
submatriz = ventas_2d[0:2, 0:3]
print("\n4. Submatriz (T1-T2, Primeras 3 regiones):\n", submatriz)

# 5. Agregaciones por eje (axis)
# axis=1 suma a lo ancho (las regiones de un mismo trimestre)
# axis=0 suma a lo alto (los trimestres de una misma región)
totales_trimestre = np.sum(ventas_2d, axis=1)
totales_region = np.sum(ventas_2d, axis=0)
print("\n5. Totales por trimestre:", totales_trimestre)
print("   Totales por región:", totales_region)

# 6. Previsión T4 (+6% sobre T3) y apilamiento (vstack)
t4_prevision = ventas_2d[2, :] * 1.06
ventas_con_t4 = np.vstack((ventas_2d, t4_prevision))
print("\n6. Matriz con T4 añadido:\n", ventas_con_t4.round(2))

# 7. Región con mayor facturación acumulada
nombres_regiones = ["Norte", "Sur", "Este", "Oeste", "Centro"]
indice_max = np.argmax(totales_region)
mejor_region = nombres_regiones[indice_max]
print("\n7. Región con mayor facturación global:", mejor_region)
```

## Ejercicio 11 — Estadística descriptiva sobre telemetría simulada

```python
import numpy as np

# 1. Generar arrays de 500 elementos con semilla fija
rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

# 2. Estadística descriptiva básica
media_temp = np.mean(temperatura)
mediana_temp = np.median(temperatura)
std_temp = np.std(temperatura)

print("2. Estadísticas de Temperatura:")
print(f"   Media  : {media_temp:.2f} °C")
print(f"   Mediana: {mediana_temp:.2f} °C")
print(f"   Desv   : {std_temp:.2f} | Min: {np.min(temperatura):.2f} | Max: {np.max(temperatura):.2f}")

# 3. Percentiles de vibración
percentiles_vib = np.percentile(vibracion, [25, 50, 75, 95])
print("\n3. Percentiles de vibración (25, 50, 75, 95):")
print("  ", np.round(percentiles_vib, 2))

# 4. Apilar arrays y verificar shape
matriz_telemetria = np.column_stack((temperatura, vibracion))
print("\n4. Shape tras apilar (column_stack):", matriz_telemetria.shape)

# 5. Matriz de correlación
matriz_corr = np.corrcoef(temperatura, vibracion)
print("\n5. Matriz de correlación:\n", np.round(matriz_corr, 3))

# 6. Cálculo de umbral (media + 2*desviación)
umbral = media_temp + (2 * std_temp)
superan_umbral = np.sum(temperatura > umbral)
porcentaje = (superan_umbral / len(temperatura)) * 100

print(f"\n6. Umbral calculado (Media + 2*Std): {umbral:.2f} °C")
print(f"   Lecturas que lo superan: {superan_umbral} ({porcentaje:.1f}%)")
```

**Media vs Mediana:** Como la media y la mediana tienen valores casi idénticos, indica que la distribución de las temperaturas es simétrica (sin valores extremos atípicos que tiren de la media hacia un lado).

**Correlación:** Los valores fuera de la diagonal principal son muy cercanos a 0, lo que explica que no existe relación lineal matemática entre la temperatura y la vibración generadas.

**Uso de la semilla:** Fijar la semilla (`default_rng(2024)`) es imprescindible porque congela el algoritmo de aleatoriedad en un punto exacto. Sin ella, los números cambiarían cada vez que ejecutamos la celda, haciendo imposible que un auditor o compañero replique y valide nuestras conclusiones exactas.

## Ejercicio 12 — Diccionarios simples y anidados

```python
# 1. Crear el diccionario
tarifas_envio = {
    "Norte": 4.95, 
    "Sur": 5.50, 
    "Este": 5.20, 
    "Oeste": 6.10, 
    "Centro": 3.90
}

# 2. Acceder a un valor por su clave
print("2. Tarifa Este:", tarifas_envio["Este"], "€")

# 3. Añadir nueva región y actualizar otra
tarifas_envio["Insular"] = 9.80
tarifas_envio["Oeste"] = 5.95

# 4. Eliminar y comprobar existencia
del tarifas_envio["Sur"]
comprobacion_sur = "Sur" in tarifas_envio
print("\n4. ¿Existe la región 'Sur' tras borrarla?:", comprobacion_sur)

# 5. Recorrer el diccionario con .items()
print("\n5. Tarifas actualizadas:")
for region, tarifa in tarifas_envio.items():
    print(f"   {region} -> {tarifa} €")

# 6. Crear diccionario anidado
red_logistica = {
    "Norte":  {"almacen": "ALM-NORTE", "tarifa": 4.95, "entrega_h": 24},
    "Este":   {"almacen": "ALM-ESTE",  "tarifa": 5.20, "entrega_h": 48},
    "Centro": {"almacen": "ALM-CENTRO","tarifa": 3.90, "entrega_h": 24},
}

# 7. Acceso mediante doble clave
tiempo_este = red_logistica["Este"]["entrega_h"]
print("\n7. Tiempo de entrega en el Este:", tiempo_este, "horas")

# 8. Añadir registro complejo e imprimir informe
red_logistica["Oeste"] = {"almacen": "ALM-OESTE", "tarifa": 5.95, "entrega_h": 72}

print("\n8. Informe de Red Logística:")
for region, datos in red_logistica.items():
    print(f"   [{region}] Almacén: {datos['almacen']} | Tarifa: {datos['tarifa']} € | Entrega: {datos['entrega_h']}h")
```

Un diccionario conviene cuando necesitas identificar los datos mediante una clave con significado para el negocio (como un nombre o un código) en lugar de depender de su orden numérico.
```python
# 1. Columna como Serie y como DataFrame
salario_serie = empleados["salario_base"]
salario_df = empleados[["salario_base"]]

print("1. Tipo usando un corchete   :", type(salario_serie))
print("   Tipo usando doble corchete:", type(salario_df))

# 2. Varias columnas a la vez (requiere lista, por tanto doble corchete)
seleccion_multi = empleados[["departamento", "ciudad", "salario_base"]]
print("\n2. Selección de varias columnas (primeras 2 filas):\n", seleccion_multi.head(2))

# 3. Filas de la 10 a la 15 con corchetes (posición)
# El slicing estándar no incluye el límite superior, por eso usamos 16 para llegar a la 15.
filas_10_15 = empleados[10:16]
print("\n3. Filas 10 a la 15 (ambas incluidas):\n", filas_10_15)

# 4. Extraer fila exacta con .loc (por etiqueta)
empleado_1042 = empleados.loc["EMP-1042"]
print("\n4. Fila completa del empleado EMP-1042:\n", empleado_1042)

# 5. .loc para filas y columnas específicas a la vez
# Sintaxis: loc[[lista_filas], [lista_columnas]]
dept_modalidad = empleados.loc[["EMP-1005", "EMP-1020", "EMP-1099"], ["departamento", "modalidad"]]
print("\n5. Departamento y modalidad de EMP-1005, 1020 y 1099:\n", dept_modalidad)

# 6. .iloc para navegar puramente por posiciones numéricas (0, 1, 2...)
# Filas: 0 a 3 (no incluida la 3). Columnas: índices 2 y 5.
iloc_posiciones = empleados.iloc[0:3, [2, 5]]
print("\n6. iloc para las primeras 3 filas y columnas 2 y 5:\n", iloc_posiciones)

# 7. .loc para todas las filas (usando los dos puntos :)
nombre_bonus = empleados.loc[:, ["nombre", "bonus_pct"]]
print("\n7. Todas las filas para nombre y bonus (muestra 3):\n", nombre_bonus.head(3))
```

| Método | Trabaja por... | Descripción |
| --- | --- | --- |

| **`.loc[]`** | Etiquetas (Labels) | Localiza filas y columnas explícitamente por el texto que aparece en sus índices y cabeceras. (El *slicing* aquí sí incluye el límite superior). |
| **`.iloc[]`** | Posición numérica (Index) | Localiza filas y columnas basándose estrictamente en su posición entera real, empezando por el 0, ignorando los nombres. |

## Ejercicio 15 — Columnas calculadas, agregación y exportación

```python
# 1. Crear importe_bruto
ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]

# 2. Crear importe_neto (aplicando el porcentaje de descuento)
ventas["importe_neto"] = round(ventas["importe_bruto"] * (1 - ventas["descuento_pct"] / 100), 2)

# 3. Crear ticket_medio por unidad
ventas["ticket_medio"] = ventas["importe_neto"] / ventas["unidades"]

# 4. Agrupación por región y ordenación descendente
facturacion_region = ventas.groupby("region")["importe_neto"].sum().sort_values(ascending=False)
print("4. Facturación por región:\n", facturacion_region)
print("   -> Total comprobación (debe ser 190402.48):", round(facturacion_region.sum(), 2), "€")

# 5. Agrupación por dos columnas simultáneas usando .agg()
# Pasamos un diccionario para aplicar funciones distintas a columnas distintas
agg_cat_canal = ventas.groupby(["categoria", "canal"]).agg({
    "importe_neto": "mean",  # Media de ingresos
    "unidades": "sum"        # Total de volumen
})
print("\n5. Media de ingresos y volumen total por Categoría y Canal:\n", round(agg_cat_canal, 2))

# 6. Top 5 productos (nlargest evita tener que ordenar toda la serie)
top_productos = ventas.groupby("producto")["importe_neto"].sum().nlargest(5)
print("\n6. Top 5 productos estrella:\n", top_productos)

# 7. Rellenar nulos (Imputación)
mediana_sat = ventas["satisfaccion"].median()
ventas["satisfaccion"] = ventas["satisfaccion"].fillna(mediana_sat)
ventas["canal"] = ventas["canal"].fillna("Desconocido")

print("\n7. Nulos restantes en 'satisfaccion':", ventas["satisfaccion"].isna().sum())
print("   Nulos restantes en 'canal'       :", ventas["canal"].isna().sum())

# 8. Exportar a CSV (usando ruta relativa hacia la carpeta outputs)
facturacion_region.to_csv("../outputs/facturacion_por_region.csv", header=True)
print("\n8. Archivo exportado con éxito a '../outputs/facturacion_por_region.csv'")
```

## Ejercicio 16 — Operadores de comparación y booleanos

```python
import numpy as np

# 1. Evaluaciones booleanas básicas
print("1. Evaluaciones:")
print("   2 < 3:", 2 < 3)
print("   'Norte' == 'norte':", "Norte" == "norte")
print("   True + True + False:", True + True + False)
print("   3 != 3.0:", 3 != 3.0)
print("   'Zapatilla' > 'Mochila':", "Zapatilla" > "Mochila")

# 2. Expresión booleana compuesta con AND
unidades = 6
precio = 149.0
descuento = 15

# Ambas condiciones deben ser ciertas a la vez
condicion_and = (unidades > 4) and (descuento >= 10)
print("\n2. Más de 4 uds Y descuento >= 10:", condicion_and)

# 3. Expresión booleana compuesta con OR
# Basta con que se cumpla una de las dos
condicion_or = (precio < 50) or (precio > 200)
print("3. Precio < 50 O precio > 200:", condicion_or)

# 4. Operador NOT (invierte el resultado)
negacion_and = not condicion_and
print("4. Negación del punto 2:", negacion_and)

# 5. Arrays de NumPy
temp = np.array([64.2, 88.1, 71.0, 92.5, 69.8])
vib = np.array([1.9, 3.4, 2.1, 4.0, 2.2])

# 6. Intentar evaluar arrays con el 'and' de Python
try:
    fallo = temp > 70 and vib > 3
except ValueError as e:
    print("\n6. Error generado:", type(e).__name__)
    print("   Mensaje:", e)

# 7. Resolución vectorizada correcta
solucion_1 = np.logical_and(temp > 70, vib > 3)
solucion_2 = (temp > 70) & (vib > 3)

print("\n7. Resultado con np.logical_and():", solucion_1)
print("   Resultado con operador &:        ", solucion_2)
print("   ¿Son métodos idénticos?:         ", np.array_equal(solucion_1, solucion_2))
```

**1. Razonamiento de las evaluaciones iniciales:**
* `2 < 3`: True, matemáticamente correcto.
* `"Norte" == "norte"`: False, porque Python es *case-sensitive* (distingue mayúsculas de minúsculas).
* `True + True + False`: 2, porque internamente True equivale a 1 y False a 0 (1+1+0=2).
* `3 != 3.0`: False, porque aunque sean de distinto tipo (int vs float), su valor numérico es el mismo, así que sí son iguales.
* `"Zapatilla" > "Mochila"`: True, porque en orden alfabético la 'Z' va después y por tanto tiene un valor mayor que la 'M'.

**4. Interpretación de negocio del `not`:** 
Al aplicar `not` a la condición 2, estamos filtrando exactamente lo contrario: "Pedidos que NO son ventas de alto volumen con gran descuento" (es decir, tienen 4 unidades o menos, o se les aplicó un descuento inferior al 10%).

**6. Explicación del ValueError con NumPy:**
El operador estándar `and` intenta evaluar el array completo como un único valor de verdad para tomar una decisión, pero NumPy no sabe si responder True o False cuando hay varios valores. Exige usar operaciones vectorizadas a nivel de elemento como `&` o `.logical_and()`.

## Ejercicio 18 — Filtrado avanzado de DataFrames

```python
# 1. Filtro múltiple tradicional (Online Y > 500€)
online_500 = ventas[(ventas["canal"] == "Online") & (ventas["importe_neto"] > 500)]
print("1. Pedidos Online > 500€:", len(online_500))

# 2. Filtrar múltiples valores exactos con .isin()
norte_centro = ventas[ventas["region"].isin(["Norte", "Centro"])]
print("2. Pedidos en Norte o Centro:", len(norte_centro))

# 3. Rango numérico con .between()
dto_10_20 = ventas[ventas["descuento_pct"].between(10, 20)]
print("3. Pedidos con descuento 10-20%:", len(dto_10_20))

# 4. Combinar tres condiciones
electronica_volumen = ventas[
    (ventas["categoria"] == "Electronica") & 
    (ventas["unidades"] > 2) & 
    (ventas["descuento_pct"] != 0)
]
print("4. Electrónica, >2 uds, con descuento aplicado:", len(electronica_volumen))

# 5. Invertir condición con la virgulilla ~ (No Oficina)
no_oficina = ventas[~(ventas["categoria"] == "Oficina")]
print("5. Pedidos que NO son de Oficina:", len(no_oficina))

# 6. Mismo filtro del paso 1, pero usando .query()
query_online_500 = ventas.query("canal == 'Online' and importe_neto > 500")
son_iguales = len(online_500) == len(query_online_500)
print("\n6. Resultados idénticos entre corchetes y query?:", son_iguales)

# 7. Filtrado y ordenación encadenada sobre 'empleados'
ingenieros_senior = empleados[
    (empleados["departamento"] == "Ingenieria") & 
    (empleados["modalidad"] == "Remoto") & 
    (empleados["antiguedad_anios"] > 5)
].sort_values(by="salario_base", ascending=False)

print("\n7. Top 3 Ingenieros Senior en Remoto (por salario):\n", ingenieros_senior[["nombre", "salario_base"]].head(3))

# 8. Exportar a CSV (respeta el índice 'id_empleado' automáticamente)
ingenieros_senior.to_csv("../outputs/ingenieria_remoto_senior.csv", header=True)
print("\n8. Exportado con éxito a '../outputs/ingenieria_remoto_senior.csv'")
```

La sintaxis con `.query()` es mucho más legible y similar a una consulta de base de datos (SQL), ya que nos evita tener que escribir el nombre del DataFrame antes de cada columna (`ventas["..."]`) y nos permite usar las palabras naturales `and` / `or` en lugar de los símbolos `&` / `|` con su estricto sistema de paréntesis.

## Ejercicio 19 — Bucles sobre estructuras de datos

```python
import numpy as np

# Recuperamos los datos de partida para no depender de celdas antiguas
regiones = ["Norte", "Sur", "Este", "Oeste", "Centro"]
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]
tarifas_envio = {"Norte": 4.95, "Este": 5.20, "Oeste": 5.95, "Centro": 3.90, "Insular": 9.80}
np_precios = np.array([59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00])

# 1. Bucle while (Reposición de stock)
print("1. Reposición de stock:")
stock = 12
lote = 0
while stock < 60:
    lote = lote + 1
    stock = stock + 8
    print("   Lote", lote, "-> stock:", stock)

# 2. Bucle for sobre lista simple
print("\n2. Regiones en mayúsculas:")
for region in regiones:
    print("  ", region.upper())

# 3. enumerate para listas (índice + valor)
print("\n3. Regiones enumeradas:")
for i, region in enumerate(regiones, start=1):
    print(f"   Región {i}: {region}")

# 4. Lista de listas (Desempaquetado directo en la cabecera)
print("\n4. Valor del inventario:")
for almacen, producto, unidades, coste in inventario:
    print(f"   {almacen} | {producto} | {round(unidades * coste, 2)} €")

# 5. Diccionario con .items()
print("\n5. Tarifas de envío:")
for region, tarifa in tarifas_envio.items():
    print(f"   {region}: {tarifa} €")

# 6. Array NumPy con iteración simple
print("\n6. Precios con recargo (bucle):")
for precio in np_precios:
    print("  ", round(precio * 1.04, 2), "€")
```
