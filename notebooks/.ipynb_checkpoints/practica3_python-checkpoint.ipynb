## Ejercicio 1 — Catálogo de productos como lista

```python
# Partimos del catálogo base
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# 1. Total de elementos y deducción de productos
# Como la estructura intercala 1 nombre y 1 precio por producto, la cantidad real de artículos es la mitad de la longitud total.
total_elementos = len(catalogo)
num_productos = total_elementos // 2
print("1. Total elementos:", total_elementos, "-> Productos reales:", num_productos)

# 2. Separar nombres y precios mediante slicing
# Extraemos saltando de 2 en 2. Los nombres empiezan en la primera posición (0) y los precios en la segunda (1).
nombres = catalogo[0::2]
precios = catalogo[1::2]
print("\n2. Lista de nombres:", nombres)
print("2. Lista de precios:", precios)
print("   Comprobación precios (debe sumar 914.40):", sum(precios))

# 3. Obtener el precio del "Robot Aspirador" dinámicamente
# Evitamos escribir el índice a mano ('hardcoding') por si la lista cambia de orden en el futuro.
# Sabemos por la estructura de negocio que el precio siempre es la posición inmediatamente posterior al nombre (+1).
idx_robot = catalogo.index("Robot Aspirador")
precio_robot = catalogo[idx_robot + 1]
print("\n3. Precio del Robot Aspirador:", precio_robot, "€")

# 4. Últimos tres productos con índices negativos
# Consultamos la cola de la lista usando -3 para traer desde el antepenúltimo elemento hasta el final.
ultimos_tres_nombres = nombres[-3:]
print("\n4. Últimos tres nombres:", ultimos_tres_nombres)

# 5. Precio medio del catálogo
# Calculamos el promedio sumando el total y dividiendo por las líneas de producto, redondeando para formato moneda.
precio_medio = round(sum(precios) / len(precios), 2)
print("\n5. Precio medio del catálogo:", precio_medio, "€")
```

## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

# 1. Producto de la cuarta fila
# Navegamos a la fila 4 (índice 3) y luego a la columna producto (índice 1).
producto_cuarta = inventario[3][1]
print("1. Producto de la cuarta fila:", producto_cuarta)

# 2. Lista valor_stock con bucle for
# Transformamos el inventario creando una nueva estructura enfocada en el valor económico.
valor_stock = []
for fila in inventario:
    producto = fila[1]
    valor_linea = round(fila[2] * fila[3], 2)
    valor_stock.append([producto, valor_linea])

print("\n2. Valor stock por línea:")
for item in valor_stock:
    print("   ", item)

# 3. Valor inmovilizado total
# Creamos una variable acumuladora en 0 y le sumamos el valor de cada producto
valor_total = 0
for item in valor_stock:
    valor_total = valor_total + item[1]

print("\n3. Valor inmovilizado total:", valor_total, "€")


# 4. Slicing contiguo para ALM-SUR
# Extraemos la sublista asumiendo agrupación visual
inventario_sur = inventario[2:4]

# Calculamos el valor de este almacén paso a paso
valor_sur = 0
for fila in inventario_sur:
    unidades = fila[2]
    coste = fila[3]
    valor_sur = valor_sur + (unidades * coste)

# Redondeamos al final
valor_sur = round(valor_sur, 2)
print("4. Valor inmovilizado ALM-SUR:", valor_sur, "€")
```

Si los registros de un mismo almacén dejan de estar en posiciones consecutivas, el slicing extraería filas incorrectas basándose únicamente en su orden numérico y no en su contenido. Por ello, el slicing no es una estrategia de filtrado robusta; en entornos reales se deben utilizar condiciones lógicas que evalúen el dato de cada fila sin depender de su ubicación.

## Ejercicio 3 — Manipulación destructiva y copias de listas

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00]

# 1. Asignación directa y modificación (el error típico)
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90
print("1. Lista original :", precios_originales)
print("   Lista rebajada :", precios_rebajados)
print("   Explicación: Al usar '=', no creamos una lista nueva, solo una 'etiqueta' que apunta a la misma lista. Modificar una altera la otra.")

# 2. Restaurar original y crear copias reales e independientes
precios_originales[0] = 59.90  # Restauramos el valor inicial

copia_con_list = list(precios_originales)
copia_con_slicing = precios_originales[:]

# Demostramos que son independientes
copia_con_list[0] = 10.00
print("\n2. Original tras cambiar la copia:", precios_originales)
print("   Copia independiente cambiada:", copia_con_list)

# 3. Rebaja del 10% iterando por posiciones (índices)
for i in range(len(copia_con_slicing)):
    nuevo_precio = copia_con_slicing[i] * 0.90
    copia_con_slicing[i] = round(nuevo_precio, 2)
    
print("\n3. Copia con 10% de rebaja:", copia_con_slicing)

# 4. Añadir e insertar productos en la lista rebajada
copia_con_slicing.extend([99.90, 120.00])
copia_con_slicing.insert(2, 75.50)
print("\n4. Tras añadir e insertar nuevos precios:", copia_con_slicing)

# 5. Eliminar elementos
del copia_con_slicing[4]
descartado = copia_con_slicing.pop()
print("\n5. Lista tras borrar posición 4 y el último:", copia_con_slicing)
print("   Valor extraído y guardado:", descartado)

# 6. Comprobar que viven en espacios de memoria distintos
print("\n6. Identificadores en memoria (id):")
print("   ID Original      :", id(precios_originales))
print("   ID Copia list()  :", id(copia_con_list))
print("   ID Copia slicing :", id(copia_con_slicing))
```

## Ejercicio 4 — Cola de pedidos pendientes

```python
cola = ["PED-10021", "PED-10007", "PED-10044", "PED-10012", "PED-10033", "PED-10008"]

# 1. Invertir con slicing negativo
cola_invertida = cola[::-1]
print("1. Cola invertida:", cola_invertida)

# 2. Ordenar con sorted() sin modificar la original
cola_alfabetica = sorted(cola)
print("\n2. Copia ordenada :", cola_alfabetica)
print("   Original intacta:", cola)

# 3. Ordenar la original
cola.sort(reverse=True)
print("\n3. Original ordenada en sitio:", cola)

# 4. Extraer y atender los dos primeros pedidos
atendidos = []
atendidos.append(cola.pop(0))
atendidos.append(cola.pop(0))
print("\n4. Pedidos atendidos:", atendidos)

# 5. Insertar pedido urgente al principio
cola.insert(0, "PED-99999")
print("\n5. Cola actual con urgencia:", cola)

# 6. Informe final
num_atendidos = len(atendidos)
num_pendientes = len(cola)
total_pedidos = num_atendidos + num_pendientes
porcentaje_avance = round((num_atendidos / total_pedidos) * 100, 1)

print("\n6. --- INFORME FINAL ---")
print(f"   Atendidos: {num_atendidos}")
print(f"   Pendientes: {num_pendientes}")
print(f"   Avance: {porcentaje_avance} %")
```

## Ejercicio 5 — Funciones propias para el cálculo de importes

```python
# 1 y 4. Definir calcular_importe con docstring
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el importe neto de un pedido aplicando un descuento porcentual.
    Devuelve el importe final redondeado a dos decimales.
    """
    bruto = unidades * precio_unitario
    neto = bruto * (1 - (descuento_pct / 100))
    return round(neto, 2)

# 2 y 4. Definir resumen_pedido con docstring
def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el desglose de un pedido.
    Devuelve una tupla con (importe_bruto, ahorro, importe_neto).
    """
    bruto = unidades * precio_unitario
    ahorro = bruto * (descuento_pct / 100)
    neto = bruto - ahorro
    return (round(bruto, 2), round(ahorro, 2), round(neto, 2))

# Mostrar la ayuda de la función
help(calcular_importe)

# 3. Llamada por palabra clave y desempaquetado de tupla
bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print("Desglose -> Bruto:", bruto, "€ | Ahorro:", ahorro, "€ | Neto:", neto, "€")

# 5. Cálculo sobre la lista de pedidos
pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = []
for ped in pedidos:
    # ped[0] son unidades, ped[1] es precio, ped[2] es descuento
    importe_calculado = calcular_importe(ped[0], ped[1], ped[2])
    importes.append(importe_calculado)

print("\nLista de importes :", importes)
print("Número de pedidos :", len(importes))
print("Importe máximo    :", max(importes), "€")
print("Importe mínimo    :", min(importes), "€")
print("Importes ordenados:", sorted(importes))
```

## Ejercicio 6 — Normalización de códigos SKU

```python
skus_crudos = [
    "  elc-0012-es ", "HOG-0045-ES", "dep-0003-pt  ",
    "ofi-0021-es", "  ELC-0012-ES", "hog-0099-fr ",
]

# 1. Limpiar espacios y convertir a mayúsculas
skus_limpios = []
for sku in skus_crudos:
    skus_limpios.append(sku.strip().upper())
print("1. SKUs limpios:", skus_limpios)

# 2. Contar apariciones de "ELC-0012-ES"
conteo_elc = skus_limpios.count("ELC-0012-ES")
print("\n2. Apariciones de ELC-0012-ES:", conteo_elc)

# 3. Obtener posición de "OFI-0021-ES"
pos_ofi = skus_limpios.index("OFI-0021-ES")
print("3. Posición de OFI-0021-ES:", pos_ofi)

# 4. Separar componentes (familia, número, país)
skus_separados = []
for sku in skus_limpios:
    skus_separados.append(sku.split("-"))
print("\n4. SKUs separados:")
for partes in skus_separados:
    print("   ", partes)

# 5. Filtrar por país "ES" y reemplazar guiones por puntos
skus_es = []
for sku in skus_limpios:
    if sku.endswith("-ES"):
        skus_es.append(sku.replace("-", "."))
print("\n5. SKUs de España modificados:", skus_es)

# 6. Unir en una cadena única separada por " | "
cadena_unica = " | ".join(skus_es)
print("\n6. Cadena única unida:", cadena_unica)
```
